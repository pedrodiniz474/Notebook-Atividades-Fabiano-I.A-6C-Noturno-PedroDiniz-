```python
#20. Matriz Transposta
# Dada uma matriz (lista de listas), crie sua matriz transposta

#Criação da matriz
matriz = [[1,2,3],[4,5,6]]
#Criação de uma lista vazia para a matriz transposta
transposta = []
#Percorre as colunas da matriz original
for coluna in range(len(matriz[0])):
  #Criação de nova linha vazia
  nova_linha = []
   #Percorre as linhas da matriz original
  for linha in range(len(matriz)):
  #Adiciona o valor trocando linha por cluna
    nova_linha.append(matriz[linha][coluna])
  #Adiciona a nova linha na matriz transposta
  transposta.append(nova_linha)
  #Saida da matriz transposta
  print(transposta)

  #justificativa: usei 2 for um dentro do outro, o de fora anda pelas colunas e o de dentro anda pelas linhas, pegando o valor matriz[linha][coluna] e colocando no lugar trocado
```

```python
#21.Busca Binária
# Implemente a busca binária para encontrar um elemento em uma
#lista ordenada.

#Lista ordenada
lista = [1,3,5,7,9,11,13]
#Usuário irá digitar o número que sera buscado
alvo = int(input("Digite um número que quer buscar: "))
#Marca o inicio da busca
inicio = 0
#Marca o fim da busca
fim = len(lista) - 1
#Variável para saber se achou o número
encontrado = False
#Repete enquanto não encontrar e o intervalo for válido
while inicio <= fim and not encontrado:
#Calcula a posição do meio
  meio = (inicio + fim) // 2
#Verifica se o valor do meio é o q se procura
  if lista[meio] == alvo:
    #marca que encontrou
    encontrado = True
#Se o valor do meio for menos que o alvo
  elif lista[meio] < alvo:
#Move o início para depois do meio
    inicio = meio + 1
#se não
  else:
  #Move o fim para antes do meio
    fim = meio - 1
#Verifica se encontrou o número
if encontrado:
  #Saida que encontrou
  print("Número encontrado na lista")
#Caso não tenha encontrado
else:
  #Saida que não encontrou
  print("Número não está na lista")


#Justificativaa: a cada volta do while, olho o número do meio da lista. Se for igual ao que eu quero, achei. Se for menor, procuro na metade de cima. Se for maior, procuro na metade de baixo
```

```python
#22. Criptografia de Cifra de César
# Crie uma função para criptografar uma mensagem usando a cifra de
#César.

#Cria uma função para cifrar a mensagem
def cifrar (mensagem, deslocamento):
  #Cria uma string vazia para guardar o resultado
  resultado = ""
  #Percorre cada letra da mensagem
  for letra in mensagem:
    #Verifica se é uma letra do alfabeto
    if letra.isalpha():
      #Calcula a nova letra deslocada
      nova_letra = chr((ord(letra.lower()) - 97 + deslocamento) % 26 + 97)
      #Adiciona a nova letra no resultado
      resultado = resultado + nova_letra
    #Caso nao seja letra
    else:
      #Mantem o caractere igual
      resultado = resultado + letra
  #Devolve a mensagem cifrada
  return resultado
#Usuario vai digitar a mensagem para cifrar
mensagem = input("Digite a mensagem: ")
#Pede o deslocamento
deslocamento = int(input("Digite o deslocamento: "))
#Mostra a mensagem cifrada
print(cifrar(mensagem, deslocamento))

#Justificativa: Usei ord() para pegar o número de cada letra e chr() para voltar de número para letra. O %26 garante que a letra percorre o alfabeto novamente depois que passa da letra z. Caracteres que não são letra ficam iguais.
```

```python
#23. Descriptografar Cifra de César
# Crie uma função para descriptografar uma mensagem cifrada com a
#cifra de César.

#Função para descriptografar a mensagem
def decifrar(mensagem, deslocamento):
  #String vazia para o resultado
  resultado = ""
  #Percorre cada letra da mensagem
  for letra in mensagem:
    #Verifica se é uma letra
    if letra.isalpha():
      #Calcula a letra original
      nova_letra = chr((ord(letra.lower()) - 97 - deslocamento) % 26 + 97)
      #Adiciona a letra no resultado
      resultado = resultado + nova_letra
    #Caso não seja letra
    else:
      #Mantem o caractere igual
      resultado = resultado + letra
  #Devolve a mensagem descriptograda
  return resultado
#Pede pro usuario a mensagem cifrada
mensagem = input("Digite a mensagem cifrada: ")
#Pede o deslocamento
deslocamento = int(input("Digite o deslocamento: "))
#saida da mensagem original
print(decifrar(mensagem, deslocamento))

#justificativa: usei a mesma logica do exercicio passada de cifrar, só que agora eu subtraio o deslocamento em vez de somar, pra voltar a letra pro lugar original.
```

```python
#24.Validação de CPF
#Implemente uma função para validar um CPF brasileiro.

#Usuario digita o cpf
cpf = input("Digite o CPF (somente números): ")
#Cria uma lista vazia para guardar os números
numeros = []
#Percorre cada caractere do CPF
for digito in cpf:
    #Transforma o caractere em número e adiciona na lista
    numeros.append(int(digito))

#Variável para a primeira soma
soma1 = 0
#Percorre os 9 primeiros números
for i in range(9):
    #Multiplica cada número pelo peso
    soma1 = soma1 + numeros[i] * (10 - i)
#Calcula o primeiro dígito verificador
resto1 = (soma1 * 10) % 11
#Caso o resto seja 10
if resto1 == 10:
    #O dígito vira 0
    resto1 = 0

#Variável para a segunda soma
soma2 = 0
#Percorre os 10 primeiros números
for i in range(10):
    # Multiplica cada número pelo peso
    soma2 = soma2 + numeros[i] * (11 - i)
#Calcula o segundo dígito verificador
resto2 = (soma2 * 10) % 11
#Caso o resto seja 10
if resto2 == 10:
    # O dígito vira 0
    resto2 = 0

#Verifica o tamanho e se os dígitos calculados batem com os do CPF
if len(cpf) == 11 and resto1 == numeros[9] and resto2 == numeros[10]:
    #Mostra que é válido
    print("CPF válido")
#Caso contrário
else:
    #Mostra que é inválido
    print("CPF inválido")

#Justificativa: Transformo o CPF numa lista de números. Depois calculo os dois dígitos verificadores usando a fórmula oficial do CPF (multiplicar cada número por um peso e tirar o resto da divisão por 11). No final, num único if, confiro se o CPF tem 11 dígitos e se os dois dígitos calculados batem com os do CPF digitado.
```

```python
#25. Classificação de Números
#Classifique uma lista de números em três categorias: positivos,
#negativos e zeros.

#Cria uma lista de números
numeros = [4, -3, 0, 8, -7, 0, 2]
#Lista para guardar números positivos
positivos = []
#Lista para guardar números negativos
negativos = []
#Lista para guardar os zeros
zeros = []
#Percorre cada número da lista
for numero in numeros:
    #Verifica se o número é positivo
    if numero > 0:
        #Adiciona na lista de positivos
        positivos.append(numero)
    #Verifica se o número é negativo
    elif numero < 0:
        #Adiciona na lista de negativos
        negativos.append(numero)
    #Caso o número seja zero
    else:
        #Adiciona na lista de zeros
        zeros.append(numero)
#Mostra os números positivos
print(f"Positivos: {positivos}")
#Mostra os números negativos
print(f"Negativos: {negativos}")
#Mostra os zeros
print(f"Zeros:{zeros}")

#Justificativa: Criei três listas vazias e, com um for e um if/elif/else, fui colocando cada número na lista certa.
```

```python
#30.Jogo da Forca
#Desenvolva um jogo simples da forca onde o usuário deve adivinhar
#uma palavra.

#Palavra secreta do jogo
palavra = "lol"
#Quantidade de tentativas que o jogador tem
tentativas = 6
#Lista para guardar as letras já descobertas
letras_descobertas = []
#Lista para guardar as letras erradas
letras_erradas = []

#Repete enquanto ainda houver tentativas
while tentativas > 0:
    #Cria uma string vazia para mostrar a palavra
    escondida = ""
    #Percorre cada letra da palavra secreta
    for letra in palavra:
        #Verifica se a letra já foi descoberta
        if letra in letras_descobertas:
            #Mostra a letra descoberta
            escondida = escondida + letra
        #Caso a letra ainda não tenha sido descoberta
        else:
            #Mostra um traço no lugar da letra
            escondida = escondida + "_"
    #Saida mostrando a palavra com as letras descobertas
    print(escondida)

    #Verifica se todas as letras foram descobertas
    if escondida == palavra:
        #Mostra mensagem de vitória
        print("Você venceu!")
        #Sai do loop porque o jogo acabou
        break

    #Pede uma letra ao jogador
    chute = input("Digite uma letra: ")
    #Verifica se a letra está na palavra
    if chute in palavra:
        #Adiciona a letra nas descobertas
        letras_descobertas.append(chute)
    #Caso a letra não esteja na palavra
    else:
        #Adiciona a letra nas erradas
        letras_erradas.append(chute)
        #Diminui uma tentativa
        tentativas = tentativas - 1
        #Mostra tentativas restantes
        print(f"Letra errada! Tentativas restantes: {tentativas}")

#Verifica se as tentativas acabaram
if tentativas == 0:
    #Saida da palavra correta
    print(f"Você perdeu! A palavra era: {palavra}")

#Justificativa: A palavra secreta é fixa, e a cada rodada mostro a palavra com traços nos lugares ainda não descobertos. Se o jogador errar, perde uma tentativa; se descobrir tudo, ganha.
```

```python
#31.Ordenação por Bolha
#Implemente o algoritmo de ordenação por bolha (Bubble Sort).

#Lista de números para ordenar
numeros = [5, 3, 8, 1, 9, 2]
#Guarda o tamanho da lista
tamanho = len(numeros)
#Repete várias vezes para percorrer a lista
for i in range(tamanho):
    #Percorre os pares de números vizinhos
    for j in range(tamanho - 1):
        #Verifica se o número está fora de ordem
        if numeros[j] > numeros[j + 1]:
            #Guarda o valor temporariamente
            temp = numeros[j]
            #Troca a posição dos números
            numeros[j] = numeros[j + 1]
            #Coloca o valor guardado na nova posição
            numeros[j + 1] = temp
#Mostra a lista ordenada
print(numeros)

#Justificativa: Uso dois for: o de fora repete várias vezes e o de dentro compara cada número com o vizinho. Se estiverem na ordem errada, eu troco os dois usando uma variável temporária (temp).
```

```python
#37. Funções Recursivas
#Implemente uma função recursiva para calcular o fatorial de um
#número.

#Cria a função recursiva para calcular o fatorial
def fatorial(numero):
    #Caso base: fatorial de 0 é 1
    if numero == 0:
        #Retorna 1
        return 1
    #Caso contrário
    else:
        #Chama a função de novo com um número menor
        return numero * fatorial(numero - 1)

#Pede o número ao usuário
numero = int(input("Digite um número: "))
#Mostra o resultado do fatorial
print(f"O fatorial é: {fatorial(numero)}")

#justificativa: A função fatorial chama ela mesma com um número menor até chegar no caso base (numero == 0), que já sabe a resposta direto (1).
```

```python
#46. Calcular Média e Desvio Padrão
# Calcule a média e o desvio padrão de uma lista de números.

#Lista de números de exemplo
numeros = [4, 8, 15, 24, 31, 84]
#Calcula a média da lista
media = sum(numeros) / len(numeros)

#Variável para somar as diferenças ao quadrado
soma_diferencas = 0
#Percorre cada número da lista
for numero in numeros:
    #Soma a diferença ao quadrado
    soma_diferencas = soma_diferencas + (numero - media) ** 2

#Calcula a variância
variancia = soma_diferencas / len(numeros)
#Calcula o desvio padrão tirando a raiz quadrada
desvio_padrao = variancia ** 0.5

#Mostra a média
print(f"Média: {media:.2f}")
#Mostra o desvio padrão
print(f"Desvio padrão: {desvio_padrao:.2f}")

#Justificativa: Primeiro calculo a média normal. Depois, para cada número, calculo a diferença dele para a média, elevo ao quadrado e somo tudo. Isso dividido pela quantidade de números dá a variância, e a raiz quadrada da variância é o desvio padrão.
```

```python
#48.Ordenação por Inserção
# Implemente o algoritmo de ordenação por inserção (Insertion Sort)

#Lista de números para ordenar
numeros = [7, 2, 5, 1, 9, 2]
#Percorre a lista a partir do segundo elemento
for i in range(1, len(numeros)):
    #Guarda o valor atual que será posicionado
    atual = numeros[i]
    #Posição anterior ao atual
    j = i - 1
    #Enquanto o valor anterior for maior que o atual
    while j >= 0 and numeros[j] > atual:
        #Move o valor maior para frente
        numeros[j + 1] = numeros[j]
        #Anda para a posição anterior
        j = j - 1
    #Coloca o valor atual na posição correta
    numeros[j + 1] = atual
#Mostra a lista ordenada
print(numeros)


#Justificativa: Para cada número da lista (a partir do segundo), eu guardo ele em atual e vou empurrando os números maiores que ele para frente, até achar o lugar certo pra colocá-lo.
```

```python
#49.Soma dos Dígitos
# Calcule a soma dos dígitos de um número inteiro fornecido.

#Pede um número ao usuário
numero = int(input("Digite um número inteiro: "))
#Garante que o número seja positivo
numero = abs(numero)
#Variável para guardar a soma dos dígitos
soma = 0
#Repete enquanto ainda houver dígitos
while numero > 0:
    #Pega o último dígito do número
    digito = numero % 10
    #Adiciona o dígito na soma
    soma = soma + digito
    #Remove o último dígito do número
    numero = numero // 10
#Mostra o resultado
print(f"Soma dos dígitos: {soma}")

#Justificativa: Uso % 10 para pegar sempre o último dígito do número, e // 10 para "cortar" esse último dígito. Repito isso até o número zerar, somando cada dígito no caminho.
```
