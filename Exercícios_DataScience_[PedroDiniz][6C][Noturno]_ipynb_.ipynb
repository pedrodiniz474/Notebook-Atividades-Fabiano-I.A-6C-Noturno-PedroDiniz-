```python
#17. Gere uma matriz 4x4 de n´umeros inteiros aleat´orios e multiplique cada
#valor por sua posi¸c˜ao na matriz.

# importação da biblioteca numpy
import numpy as np

# cria matriz 4x4 com numeros aleatorios entre 1 e 9
matriz = np.random.randint(1, 10, size=(4, 4))

# avisa que vai mostrar a matriz original
print("Matriz original:")

# mostra a matriz original
print(matriz)

# cria uma variavel para contar a posicao, comecando em 1
posicao = 1

# percorre cada linha da matriz
for linha in range(4):
    # percorre cada coluna da matriz
    for coluna in range(4):
        # multiplica o valor pela posicao atual
        matriz[linha][coluna] = matriz[linha][coluna] * posicao
        # aumenta a posicao para o proximo elemento
        posicao = posicao + 1

# avisa que vai mostrar a matriz final
print("Matriz multiplicada pela posicao:")

# mostra a matriz final
print(matriz)

# Justificativa: usei dois for um dentro do outro para percorrer cada elemento da matriz (linha por linha e coluna por coluna), multiplicando cada valor por um contador de posição que vai aumentando de 1 em 1.

```

```python
#19. Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o
#angulo entre eles (produto escalar normalizado).

# importação da biblioteca numpy
import numpy as np

# cria primeiro array com 20 numeros aleatorios
array_a = np.random.randint(1, 20, size=20)

# cria segundo array com 20 numeros aleatorios
array_b = np.random.randint(1, 20, size=20)

# calcula o produto escalar entre os dois arrays
produto_escalar = np.dot(array_a, array_b)

# calcula o tamanho do array a
norma_a = np.linalg.norm(array_a)

# calcula o tamanho do array b
norma_b = np.linalg.norm(array_b)

# calcula o cosseno do angulo entre os arrays
cosseno = produto_escalar / (norma_a * norma_b)

# calcula o angulo em radianos
angulo = np.arccos(cosseno)

# transforma o angulo de radianos para graus
angulo_graus = np.degrees(angulo)

# mostra o angulo final
print(f"O angulo entre os arrays é: {angulo_graus} graus")

# Justificativa: usei a fórmula do cosseno do ângulo entre vetores, depois arcos para achar o ângulo em radianos e converti para graus.
```

```python
#2 Exercıcios de SciPy
#1. Calcule a integral da funcao f(x) = x² sin(x) no intervalo de 0 a π

# importação do modulo integrate da biblioteca scipy
from scipy import integrate

# importação da biblioteca numpy
import numpy as np

# cria uma funcao para representar x ao quadrado vezes seno de x
def funcao(x):
    # retorna o valor da funcao
    return x**2 * np.sin(x)

# calcula a integral de 0 ate pi
resultado, erro = integrate.quad(funcao, 0, np.pi)

# mostra o resultado da integral
print(f"Resultado da integral: {resultado}")

# mostra o erro estimado
print(f"Erro estimado: {erro}")

# Justificativa: usei quad, que calcula a integral de uma função dentro de um intervalo, igual mostrado no material de apoio. Só precisei escrever a função certa dentro de def.
```

```python
#2. Resolva a equacao diferencial dy dx = x − y, com y(0) = 1, no intervalo de 0 a 10.

# importação da funcao solve_ivp da biblioteca scipy
from scipy.integrate import solve_ivp

# importação da biblioteca numpy
import numpy as np

# cria a funcao que representa a equacao diferencial
def equacao(x, y):
    # retorna x menos y
    return x - y

# resolve a equacao com y(0) = 1
solucao = solve_ivp(equacao, [0, 10], [1], t_eval=np.linspace(0, 10, 50))

# mostra os valores de x calculados
print("Valores de x:", solucao.t)

# mostra os valores de y calculados
print("Valores de y:", solucao.y[0])

# Justificativa: usei solve_ivp igual ao exemplo do material, colocando a condição inicial y(0) = 1 e pedindo 50 pontos entre 0 e 10 para acompanhar como y muda.
```

```python
#3. Encontre as raızes da funcao f(x) = x 4 − 3x 3 + 2 utilizando SciPy.

# importação da funcao fsolve da biblioteca scipy
from scipy.optimize import fsolve

# cria a funcao para encontrar a raiz
def funcao(x):
    # retorna o valor da funcao
    return x**4 - 3*x**3 + 2

# procura a raiz da funcao comecando perto de x = 1
raiz = fsolve(funcao, 1)

# mostra a raiz encontrada
print(f"Raiz encontrada: {raiz}")

# Justificativa: usei fsolve partindo de um chute inicial (x = 1) para o programa procurar sozinho onde a função vale zero.
```

```python
#4. Use a funcao de otimizacao de SciPy para maximizar f(x) = −x²+ 4x+ 1.

# importação da funcao minimize da biblioteca scipy
from scipy.optimize import minimize

# cria a funcao que queremos maximizar, ja invertida
def funcao(x):
    # inverte o sinal da funcao para poder usar minimize
    return -(-x**2 + 4*x + 1)

# procura o minimo da funcao invertida (que é o maximo da original)
resultado = minimize(funcao, 0)

# mostra o x que maximiza a funcao
print(f"O valor de x que maximiza a funcao é: {resultado.x}")

# Justificativa: o SciPy só tem função pronta para minimizar, então multipliquei a função original por -1 para poder achar o valor máximo usando minimize.
```

```python
#10. Calcule a integral dupla da fun¸c˜ao f(x, y) = xy + x² no intervalo [0, 2] para x e [0, 1] para y.

# importação do modulo integrate da biblioteca scipy
from scipy import integrate

# cria a funcao com y primeiro pois é a ordem que o dblquad pede
def funcao(y, x):
    # retorna o valor da funcao
    return x*y + x**2

# calcula a integral dupla
resultado, erro = integrate.dblquad(funcao, 0, 2, 0, 1)

# mostra o resultado
print(f"Resultado da integral dupla: {resultado}")

# mostra o erro estimado
print(f"Erro estimado: {erro}")

# Justificativa: usei dblquad, que é parecida com a quad do material mas serve para funções com duas variáveis (x e y).
```

```python
#16. Resolva a equacao diferencial de segunda ordem d²y|dx² +y = 0, com condicoes iniciais y(0) = 1 e y′ (0) = 0.

# importação da funcao solve_ivp da biblioteca scipy
from scipy.integrate import solve_ivp

# importação da biblioteca numpy
import numpy as np

# cria a funcao com o sistema de equacoes
def sistema(x, valores):
    # separa y e a derivada de y (chamada de v)
    y, v = valores
    # a derivada de y é v
    dydx = v
    # a derivada de v é -y, que vem da equacao original
    dvdx = -y
    # retorna as duas derivadas
    return [dydx, dvdx]

# resolve o sistema com y(0)=1 e y'(0)=0
solucao = solve_ivp(sistema, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 50))

# mostra os valores de x
print("Valores de x:", solucao.t)

# mostra os valores de y encontrados
print("Valores de y:", solucao.y[0])

# Justificativa: o solve_ivp só resolve equações de primeira ordem, então transformei a equação de segunda ordem em duas equações de primeira ordem, usando uma variável extra (v) para representar a derivada de y.
```

```python
#6. Plote dois graficos de linha em subplots diferentes: y1 = e (elevado a x) e y² = log(x).

# importação da biblioteca matplotlib
import matplotlib.pyplot as plt

# importação da biblioteca numpy
import numpy as np

# cria valores de x comecando em 0.1 para nao dar erro no log
x = np.linspace(0.1, 5, 100)

# calcula y1 como e elevado a x
y1 = np.exp(x)

# calcula y2 como logaritmo de x
y2 = np.log(x)

# cria o primeiro grafico
plt.subplot(2, 1, 1)

# plota o grafico de e elevado a x
plt.plot(x, y1)

# coloca o titulo do primeiro grafico
plt.title("Grafico de e^x")

# cria o segundo grafico
plt.subplot(2, 1, 2)

# plota o grafico de log(x)
plt.plot(x, y2)

# coloca o titulo do segundo grafico
plt.title("Grafico de log(x)")

# ajusta o layout para nao sobrepor os graficos
plt.tight_layout()

# mostra os graficos na tela
plt.show()

# Justificativa: Segue o exemplo de subplots do material.
```

```python
#8. Plote um grafico de linha com intervalos de confianca (sombramento) para a funcao y = sin(x).

# importação da biblioteca matplotlib
import matplotlib.pyplot as plt

# importação da biblioteca numpy
import numpy as np

# cria 100 valores de x entre 0 e 10
x = np.linspace(0, 10, 100)

# calcula y como seno de x
y = np.sin(x)

# define um valor fixo de erro para o sombreamento
erro = 0.2

# plota a linha principal
plt.plot(x, y, label='seno')

# cria a sombra em volta da linha
plt.fill_between(x, y - erro, y + erro, alpha=0.3)

# coloca o titulo do grafico
plt.title("Grafico com Intervalo de Confianca")

# coloca o nome do eixo x
plt.xlabel("Eixo X")

# coloca o nome do eixo y
plt.ylabel("Eixo Y")

# mostra a legenda
plt.legend()

# mostra o grafico na tela
plt.show()

# Justificativa: usei fill_between para pintar uma faixa em volta da linha, representando o intervalo de confiança. O alpha deixa a cor mais transparente.
```

```python
#10. Plote tres graficos de linha em um unico grafico, cada um com uma cor e estilo de linha diferentes.

# importação da biblioteca matplotlib
import matplotlib.pyplot as plt

# importação da biblioteca numpy
import numpy as np

# cria 100 valores de x entre 0 e 10
x = np.linspace(0, 10, 100)

# calcula a primeira linha como seno de x
y1 = np.sin(x)

# calcula a segunda linha como cosseno de x
y2 = np.cos(x)

# calcula a terceira linha como tangente de x
y3 = np.tan(x)

# plota a primeira linha
plt.plot(x, y1, color='blue', linestyle='-', label='seno')

# plota a segunda linha
plt.plot(x, y2, color='red', linestyle='--', label='cosseno')

# plota a terceira linha
plt.plot(x, y3, color='green', linestyle=':', label='tangente')

# coloca o titulo do grafico
plt.title("Tres Graficos de Linha")

# coloca o nome do eixo x
plt.xlabel("Eixo X")

# coloca o nome do eixo y
plt.ylabel("Eixo Y")

# limita o eixo y para o grafico ficar legivel
plt.ylim([-5, 5])

# mostra a legenda com os nomes das linhas
plt.legend()

# mostra o grafico na tela
plt.show()

# Justificativa: usei plot três vezes, cada vez com uma cor e um estilo de linha diferente (cheia, tracejada e pontilhada), e limitei o eixo y porque a tangente tem valores muito altos e bagunça o gráfico.
```

```python
#11. Crie um grafico de barras empilhadas para tres categorias com tres grupos de valores.

# importação da biblioteca matplotlib
import matplotlib.pyplot as plt

# importação da biblioteca numpy
import numpy as np

# cria a lista com as categorias
categorias = ['A', 'B', 'C']

# cria os valores do primeiro grupo
grupo1 = [3, 5, 4]

# cria os valores do segundo grupo
grupo2 = [2, 3, 5]

# cria os valores do terceiro grupo
grupo3 = [1, 2, 2]

# transforma o grupo 1 em array para poder somar
soma1 = np.array(grupo1)

# soma o grupo 1 com o grupo 2 para saber onde o grupo 3 comeca
soma2 = soma1 + np.array(grupo2)

# plota o primeiro grupo
plt.bar(categorias, grupo1, label='Grupo 1')

# plota o segundo grupo em cima do primeiro
plt.bar(categorias, grupo2, bottom=grupo1, label='Grupo 2')

# plota o terceiro grupo em cima dos outros dois
plt.bar(categorias, grupo3, bottom=soma2, label='Grupo 3')

# coloca o titulo do grafico
plt.title("Grafico de Barras Empilhadas")

# coloca o nome do eixo x
plt.xlabel("Categorias")

# coloca o nome do eixo y
plt.ylabel("Valores")

# mostra a legenda
plt.legend()

# mostra o grafico na tela
plt.show()

# Justificativa: segui a lógica do material, que empilhava dois grupos, mas com três grupos precisei somar os dois primeiros para saber a altura onde o terceiro deveria começar.
```

```python
#18. Crie um gr´afico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

# importação da biblioteca matplotlib
import matplotlib.pyplot as plt

# cria a lista com as categorias
categorias = ['A', 'B', 'C', 'D']

# cria os valores, alguns positivos e outros negativos
valores = [10, -5, 8, -3]

# cria uma lista vazia para guardar as cores
cores = []

# percorre cada valor da lista
for valor in valores:
    # verifica se o valor e positivo
    if valor >= 0:
        # adiciona a cor verde na lista
        cores.append('green')
    else:
        # adiciona a cor vermelha na lista
        cores.append('red')

# cria o grafico de barras usando as cores da lista
plt.bar(categorias, valores, color=cores)

# coloca o titulo do grafico
plt.title("Grafico de Barras Positivas e Negativas")

# coloca o nome do eixo x
plt.xlabel("Categorias")

# coloca o nome do eixo y
plt.ylabel("Valores")

# desenha uma linha no zero para facilitar a visualizacao
plt.axhline(0, color='black', linewidth=0.8)

# mostra o grafico na tela
plt.show()

# Justificativa: usei um for com if para verificar se cada valor é positivo ou negativo e escolher a cor certa, e axhline desenha uma linha no zero para ficar mais fácil ver os valores negativos.
```
